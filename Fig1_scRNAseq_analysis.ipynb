```python
%matplotlib inline

import scanpy as sc
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
from scipy import sparse

mpl.rcParams["svg.fonttype"] = "none"

# ======================
# Load data
# ======================
adata = sc.read_h5ad("Bleomycin_V3V9_RunID-run-240123-105643_ScanpyObject.h5ad")

# ======================
# Make GFP/RFP group
# ======================
# sample_name に GFP / RFP が入っている前提
adata.obs["GFP_RFP"] = adata.obs["sample_name"].astype(str)

# 必要に応じて表記を統一
adata.obs["GFP_RFP"] = adata.obs["GFP_RFP"].replace({
    "GFP+": "GFP",
    "GFP_positive": "GFP",
    "RFP+": "RFP",
    "RFP_positive": "RFP",
})

adata_sub = adata[adata.obs["GFP_RFP"].isin(["GFP", "RFP"])].copy()

print(adata_sub.obs["GFP_RFP"].value_counts())

# ======================
# DEG: GFP vs RFP
# ======================
sc.tl.rank_genes_groups(
    adata_sub,
    groupby="GFP_RFP",
    groups=["GFP"],
    reference="RFP",
    method="wilcoxon",
    n_genes=adata_sub.shape[1]
)

deg_gfp = sc.get.rank_genes_groups_df(adata_sub, group="GFP")

deg_gfp_up = (
    deg_gfp
    .dropna(subset=["names", "logfoldchanges", "pvals_adj"])
    .query("logfoldchanges > 0")
    .sort_values(["pvals_adj", "logfoldchanges"], ascending=[True, False])
)

top10_gfp_genes = deg_gfp_up["names"].head(10).tolist()

print("Top 10 GFP-up genes:")
print(top10_gfp_genes)

# ======================
# Extract expression matrix
# ======================
adata_plot = adata_sub[:, top10_gfp_genes].copy()

X = adata_plot.X

if sparse.issparse(X):
    X = X.toarray()

X = np.asarray(X)

# cells x genes -> genes x cells
mat = X.T

# ======================
# Gene-wise z-score
# ======================
gene_means = mat.mean(axis=1, keepdims=True)
gene_sds = mat.std(axis=1, keepdims=True)

gene_sds[gene_sds == 0] = 1

zmat = (mat - gene_means) / gene_sds
zmat = np.clip(zmat, -2.5, 2.5)

# ======================
# Sort cells: GFP first, then RFP
# ======================
obs = adata_plot.obs.copy()

cell_order = (
    obs.assign(
        group_order=obs["GFP_RFP"].map({"GFP": 0, "RFP": 1})
    )
    .sort_values(["group_order"])
    .index
)

cell_indices = [adata_plot.obs_names.get_loc(c) for c in cell_order]

zmat = zmat[:, cell_indices]
groups_ordered = obs.loc[cell_order, "GFP_RFP"].values

# ======================
# Plot
# ======================
fig, ax = plt.subplots(figsize=(10, 3.0))

im = ax.imshow(
    zmat,
    aspect="auto",
    interpolation="nearest",
    cmap="viridis",
    vmin=-2.5,
    vmax=2.5
)

# y-axis genes
ax.set_yticks(np.arange(len(top10_gfp_genes)))
ax.set_yticklabels(top10_gfp_genes, fontsize=8)

# x-axis: no individual cell labels
ax.set_xticks([])

# boundary between GFP and RFP
n_gfp = np.sum(groups_ordered == "GFP")
ax.axvline(n_gfp - 0.5, color="white", linewidth=1.2)

# group labels
ax.text(n_gfp / 2, len(top10_gfp_genes) + 0.4, "GFP", ha="center", va="top", fontsize=9)
ax.text(n_gfp + (len(groups_ordered) - n_gfp) / 2, len(top10_gfp_genes) + 0.4, "RFP", ha="center", va="top", fontsize=9)

# colorbar
cbar = fig.colorbar(im, ax=ax, fraction=0.025, pad=0.02)
cbar.set_label("Gene-wise z-score", rotation=270, labelpad=15)

ax.set_title("Top 10 GFP-upregulated genes", fontsize=11)
ax.tick_params(axis="both", length=0)

fig.tight_layout()
fig.savefig("GFP_up_top10_heatmap_gene_zscore.svg", bbox_inches="tight")
fig.savefig("GFP_up_top10_heatmap_gene_zscore.tiff", dpi=300, bbox_inches="tight")
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl
from scipy import sparse


# ============================================================
# User settings
# ============================================================

# adata.obs 内で GFP / RFP の分類が入っている列名
group_key = "GFP_RFP"

# 表示順
group_order = ["GFP", "RFP"]

# 下側 annotation bar の色
group_colors = {
    "GFP": "#1f77b4",   # blue
    "RFP": "#ff7f0e",   # orange
}

# raw layer を使うかどうか
# adata.raw にlog-normalized expressionが入っているなら True
use_raw = True

# z-scoreの表示範囲
# 極端な値でcolor scaleが潰れないようにclipします。
z_clip = 2.5

# colormap
cmap = "viridis"


# ============================================================
# Gene lists
# ============================================================

# Fig.1h: Top 10 GFP-upregulated genes
fig1h_genes = [
    "Cdkn1a",
    "Cdkn2b",
    "Mdm2",
    "Npc2",
    "Gas6",
    "Fn1",
    "AC134224.3",
    "AC134224.1",
    "Gpnmb",
    "Itm2b",
]

# Fig.1i: SASP-related genes
fig1i_genes = [
    "Il6",
    "Ccl2",
    "Ccl7",
    "Cxcl1",
    "Cxcl10",
    "Tgfb1",
    "Igfbp2",
    "Igfbp3",
    "Igfbp4",
    "Igfbp5",
    "Igfbp7",
    "Spp1",
    "Mmp2",
    "Mmp3",
    "Fn1",
    "Timp1",
    "Timp3",
    
]


# ============================================================
# Helper functions
# ============================================================

def select_adata_layer(adata, use_raw=True):
    """
    Use adata.raw if available and requested.
    Otherwise use adata.
    """
    if use_raw and adata.raw is not None:
        return adata.raw.to_adata()
    return adata


def find_genes_in_adata(ad, genes):
    """
    Find genes in AnnData.

    This supports three common cases:
    1. ad.var_names are gene symbols: Cdkn1a
    2. ad.var["gene_symbol"] contains gene symbols
    3. ad.var_names are formatted as GeneSymbol:ENSRNOG..., e.g. Cdkn1a:ENSRNOG0000000521
    """

    gene_to_var = {}
    var_names = pd.Index(ad.var_names.astype(str))

    # Case 1: exact match in ad.var_names
    for gene in genes:
        if gene in var_names:
            gene_to_var[gene] = gene

    # Case 2: gene_symbol column
    if "gene_symbol" in ad.var.columns:
        gene_symbols = ad.var["gene_symbol"].astype(str)
        for gene in genes:
            if gene not in gene_to_var:
                hits = ad.var_names[gene_symbols == gene].tolist()
                if len(hits) > 0:
                    gene_to_var[gene] = hits[0]

    # Case 3: var_names like "Cdkn1a:ENSRNOG0000000521"
    for gene in genes:
        if gene not in gene_to_var:
            hits = [v for v in ad.var_names.astype(str) if v.split(":")[0] == gene]
            if len(hits) > 0:
                gene_to_var[gene] = hits[0]

    genes_found = [g for g in genes if g in gene_to_var]
    genes_missing = [g for g in genes if g not in gene_to_var]

    return gene_to_var, genes_found, genes_missing


def extract_expression_df(adata, genes, use_raw=True):
    """
    Extract expression dataframe.
    Rows = cells
    Columns = genes
    """

    ad = select_adata_layer(adata, use_raw=use_raw)

    gene_to_var, genes_found, genes_missing = find_genes_in_adata(ad, genes)

    print("====================================")
    print("Requested genes:")
    print(genes)
    print("Found genes:")
    print(genes_found)
    print("Missing genes:")
    print(genes_missing)
    print("====================================")

    if len(genes_found) == 0:
        raise ValueError("No requested genes were found in adata.")

    var_ids = [gene_to_var[g] for g in genes_found]

    X = ad[:, var_ids].X

    if sparse.issparse(X):
        X = X.toarray()

    expr_df = pd.DataFrame(
        X,
        index=ad.obs_names,
        columns=genes_found
    )

    return expr_df, genes_found, genes_missing


def gene_wise_zscore(expr_df, z_clip=2.5):
    """
    Convert expression values to gene-wise z-scores.
    Each gene is standardized across cells.

    z = (expression - gene mean across cells) / gene SD across cells
    """

    z_df = expr_df.copy()

    for gene in z_df.columns:
        mean_val = z_df[gene].mean()
        sd_val = z_df[gene].std()

        if sd_val == 0 or np.isnan(sd_val):
            z_df[gene] = 0
        else:
            z_df[gene] = (z_df[gene] - mean_val) / sd_val

    z_df = z_df.clip(-z_clip, z_clip)

    return z_df


def sort_cells_by_group(z_df, adata, group_key, group_order):
    """
    Sort cells by GFP/RFP group.
    """

    if group_key not in adata.obs.columns:
        raise KeyError(
            f"{group_key} was not found in adata.obs. "
            f"Available columns are: {list(adata.obs.columns)}"
        )

    obs_df = adata.obs.loc[z_df.index].copy()
    obs_df[group_key] = obs_df[group_key].astype(str)

    detected_groups = sorted(obs_df[group_key].unique().tolist())
    print(f"Detected groups in {group_key}: {detected_groups}")

    missing_groups = [g for g in group_order if g not in detected_groups]
    if len(missing_groups) > 0:
        print(f"Warning: these groups in group_order were not found: {missing_groups}")

    obs_df[group_key] = pd.Categorical(
        obs_df[group_key],
        categories=group_order,
        ordered=True
    )

    sorted_cells = obs_df.sort_values(group_key).index
    z_df_sorted = z_df.loc[sorted_cells]
    group_series = obs_df.loc[sorted_cells, group_key].astype(str)

    return z_df_sorted, group_series


def plot_gene_heatmap_with_bottom_group_bar(
    adata,
    genes,
    title,
    output_prefix,
    group_key="GFP_RFP",
    group_order=("GFP", "RFP"),
    group_colors=None,
    use_raw=True,
    z_clip=2.5,
    cmap="viridis",
    figsize=(10, 3.0),
    gene_label_fontsize=8,
    title_fontsize=12,
):
    """
    Plot a gene-wise z-score heatmap with a bottom GFP/RFP annotation bar.

    Rows = genes
    Columns = cells
    """

    if group_colors is None:
        group_colors = {
            "GFP": "#1f77b4",
            "RFP": "#ff7f0e",
        }

    # --------------------------------------------------------
    # Extract expression
    # --------------------------------------------------------
    expr_df, genes_found, genes_missing = extract_expression_df(
        adata=adata,
        genes=genes,
        use_raw=use_raw
    )

    # --------------------------------------------------------
    # Gene-wise z-score
    # --------------------------------------------------------
    z_df = gene_wise_zscore(expr_df, z_clip=z_clip)

    # --------------------------------------------------------
    # Sort cells by group
    # --------------------------------------------------------
    z_df, group_series = sort_cells_by_group(
        z_df=z_df,
        adata=adata,
        group_key=group_key,
        group_order=list(group_order)
    )

    # Matrix for plotting: genes x cells
    plot_mat = z_df.T.values
    gene_labels = z_df.columns.tolist()

    # --------------------------------------------------------
    # Prepare bottom annotation bar
    # --------------------------------------------------------
    group_to_int = {group: i for i, group in enumerate(group_order)}

    group_ids = []
    for g in group_series:
        if g in group_to_int:
            group_ids.append(group_to_int[g])
        else:
            group_ids.append(np.nan)

    group_ids = np.array(group_ids).reshape(1, -1)

    color_list = [group_colors[g] for g in group_order]
    group_cmap = mpl.colors.ListedColormap(color_list)

    # --------------------------------------------------------
    # Figure layout
    # --------------------------------------------------------
    fig = plt.figure(figsize=figsize)

    gs = fig.add_gridspec(
        nrows=2,
        ncols=2,
        height_ratios=[12, 0.9],
        width_ratios=[30, 0.8],
        hspace=0.08,
        wspace=0.05
    )

    ax_heatmap = fig.add_subplot(gs[0, 0])
    ax_group = fig.add_subplot(gs[1, 0])
    ax_cbar = fig.add_subplot(gs[0, 1])

    # --------------------------------------------------------
    # Main heatmap
    # --------------------------------------------------------
    im = ax_heatmap.imshow(
        plot_mat,
        aspect="auto",
        interpolation="nearest",
        cmap=cmap,
        vmin=-z_clip,
        vmax=z_clip
    )

    ax_heatmap.set_title(title, fontsize=title_fontsize, pad=8)
    ax_heatmap.set_xticks([])
    ax_heatmap.set_yticks(np.arange(len(gene_labels)))
    ax_heatmap.set_yticklabels(gene_labels, fontsize=gene_label_fontsize)
    ax_heatmap.tick_params(axis="y", length=0)

    # --------------------------------------------------------
    # Colorbar
    # --------------------------------------------------------
    cbar = fig.colorbar(im, cax=ax_cbar)
    cbar.set_label("Gene-wise z-score", rotation=270, labelpad=16)

    # --------------------------------------------------------
    # Bottom group annotation bar
    # --------------------------------------------------------
    ax_group.imshow(
        group_ids,
        aspect="auto",
        interpolation="nearest",
        cmap=group_cmap,
        vmin=0,
        vmax=len(group_order) - 1
    )

    ax_group.set_yticks([])
    ax_group.set_xticks([])

    # Group labels under the bar
    group_counts = group_series.value_counts().reindex(group_order).fillna(0).astype(int)

    start = 0
    for group in group_order:
        count = group_counts[group]
        if count > 0:
            center = start + count / 2
            ax_group.text(
                center,
                1.55,
                group,
                ha="center",
                va="top",
                fontsize=9,
                rotation=90
            )
        start += count

    ax_group.set_xlabel(group_key, fontsize=10, labelpad=18)

    # --------------------------------------------------------
    # Remove frames
    # --------------------------------------------------------
    for ax in [ax_heatmap, ax_group]:
        for spine in ax.spines.values():
            spine.set_visible(False)

    # --------------------------------------------------------
    # Save as SVG
    # --------------------------------------------------------
    svg_path = f"{output_prefix}.svg"

    plt.savefig(
        svg_path,
        bbox_inches="tight",
        format="svg"
    )

    plt.show()

    return {
        "genes_found": genes_found,
        "genes_missing": genes_missing,
        "output_svg": svg_path,
    }


# ============================================================
# Fig.1h
# ============================================================

fig1h_result = plot_gene_heatmap_with_bottom_group_bar(
    adata=adata,
    genes=fig1h_genes,
    title="Top 10 GFP-upregulated genes",
    output_prefix="Fig1h_top10_GFP_upregulated_genes_heatmap",
    group_key=group_key,
    group_order=group_order,
    group_colors=group_colors,
    use_raw=use_raw,
    z_clip=z_clip,
    cmap=cmap,
    figsize=(10, 3.0),
    gene_label_fontsize=8,
    title_fontsize=12,
)

print("Fig.1h result:")
print(fig1h_result)


# ============================================================
# Fig.1i
# ============================================================

fig1i_result = plot_gene_heatmap_with_bottom_group_bar(
    adata=adata,
    genes=fig1i_genes,
    title="SASP-related genes",
    output_prefix="Fig1i_SASP_related_genes_heatmap",
    group_key=group_key,
    group_order=group_order,
    group_colors=group_colors,
    use_raw=use_raw,
    z_clip=z_clip,
    cmap=cmap,
    figsize=(10, 4.2),
    gene_label_fontsize=8,
    title_fontsize=12,
)

print("Fig.1i result:")
print(fig1i_result)
```
